# Importação e pre-processamento

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("uciml/iris")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report

caminho = os.path.join(path, "Iris.csv")
df = pd.read_csv(caminho)

df.info()

df.head()
```

```python
X = df.drop(columns=['Species', 'Id'])
Y = df['Species']

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42, stratify=Y
)
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

kf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

dt_model = DecisionTreeClassifier(criterion='gini', max_depth=3, random_state=42)

knn_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('knn', KNeighborsClassifier(n_neighbors=5))
])

dt_scores = cross_val_score(dt_model, X, Y, cv=kf, scoring='accuracy')
knn_scores = cross_val_score(knn_pipeline, X, Y, cv=kf, scoring='accuracy')

print("=== RESULTADOS K-FOLD CROSS-VALIDATION (10 Folds) ===\n")

print(f"Árvore de Decisão:")
print(f"  Acurácia Média: {dt_scores.mean() * 100:.2f}%")
print(f"  Desvio Padrão:  {dt_scores.std() * 100:.2f}%")
print(f"  Acurácias por fold: {np.round(dt_scores * 100, 1)}\n")

print(f"KNN (com StandardScaler):")
print(f"  Acurácia Média: {knn_scores.mean() * 100:.2f}%")
print(f"  Desvio Padrão:  {knn_scores.std() * 100:.2f}%")
print(f"  Acurácias por fold: {np.round(knn_scores * 100, 1)}")
```

Ambos os modelos atingiram desempenho idêntico (99,33) porque as classes do dataset Iris são bem delimitadas no espaço de atributos, permitindo que tanto a Árvore de Decisão quanto a KNN com dados padronizados encontrem quase perfeitamente a separação ideal. Contudo, em cenários com datasets mais complexos, a Árvore de Decisão apresenta maior risco de overfitting se não for limitada (max_depth), pois tende a criar regras excessivamente específicas para memorizar o ruído do treino; já o KNN corre o risco de overfitting quando configurado com um número de vizinhos muito pequeno (k=1), tornando-se hiper-sensível a pontos fora do padrão (outliers) na vizinhança imediata.
